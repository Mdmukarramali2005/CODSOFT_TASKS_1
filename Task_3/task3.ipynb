```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("cleaned_car_dataset.csv")
df.head()
```

```python
## Visualization 1: Top 10 Car Brands
```

```python
plt.figure(figsize=(10,6))

top_make = df['Make'].value_counts().head(10)

sns.barplot(
    x=top_make.index,
    y=top_make.values,
    palette='viridis'
)

plt.title("Top 10 Car Brands")
plt.xlabel("Car Brand")
plt.ylabel("Number of Cars")

plt.xticks(rotation=45)

plt.show()
```

```python
# Insight

# - Toyota, Hyundai, and Honda are among the most frequently listed car brands.
# - These brands dominate the dataset, indicating their strong presence in the used car market.
```

```python
## Visualization 2: Selling Price Distribution
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df['Selling_Price'],
    bins=30,
    kde=True,
    color='skyblue'
)

plt.title("Distribution of Selling Prices")
plt.xlabel("Selling Price")
plt.ylabel("Frequency")

plt.show()
```

```python
# Insight

# - Most cars are concentrated in the mid-price range.
# - A few high-priced cars appear as outliers.
```

```python
## Visualization 3: Mileage vs Selling Price
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x='Mileage',
    y='Selling_Price',
    data=df,
    hue='Fuel_Type'
)

plt.title("Mileage vs Selling Price")
plt.xlabel("Mileage")
plt.ylabel("Selling Price")

plt.show()
```

```python
# Insight

# - Cars with lower mileage generally have higher selling prices.
# - Mileage influences the resale value of a vehicle.
```

```python
## Visualization 4: Fuel Type Distribution
```

```python
fuel = df['Fuel_Type'].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    fuel,
    labels=fuel.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Fuel Type Distribution")

plt.show()
```

```python
#  Insight

# - Petrol and Diesel vehicles account for the majority of listings.
# - Electric and Hybrid vehicles represent a smaller share.
```

```python
## Visualization 5: Average Selling Price by Year
```

```python
avg_price = df.groupby('Year')['Selling_Price'].mean()

plt.figure(figsize=(12,6))

plt.plot(
    avg_price.index,
    avg_price.values,
    marker='o',
    linewidth=2
)

plt.title("Average Selling Price by Year")
plt.xlabel("Manufacturing Year")
plt.ylabel("Average Selling Price")

plt.grid(True)

plt.show()
```

```python
#  Insight

# - Average selling prices generally increase for newer manufacturing years.
# - Newer cars retain higher resale value than older models.
```

```python
## Visualization 6: Correlation Heatmap
```

```python
plt.figure(figsize=(10,8))

corr = df.select_dtypes(include='number').corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
#  Insight

# - Selling Price has noticeable relationships with several numerical features.
# - The heatmap helps identify which variables are positively or negatively correlated.
```
