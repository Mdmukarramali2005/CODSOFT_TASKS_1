```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("cleaned_car_dataset.csv")
```

```python
df
```

```python
df.head()
```

```python
df.describe()
```

```python
df.describe(include="object")
```

```python
corr = df.corr(numeric_only=True)
```

```python
corr
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
df.hist(figsize=(15, 10))
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df["Selling_Price"])
plt.title("Selling Price Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="Mileage", y="Selling_Price", data=df)
plt.title("Mileage vs Selling Price")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="Fuel_Type", data=df)
plt.title("Fuel Type Distribution")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="Body_Type", data=df)
plt.title("Body Type Distribution")
plt.xticks(rotation=45)
plt.show()
```

```python
#  Conclusion

# The Exploratory Data Analysis (EDA) revealed important patterns in the used car dataset.
# Vehicle age and mileage have a significant impact on selling price.
# Newer cars generally have higher prices, while cars with higher mileage tend to have lower prices.
# Petrol vehicles dominate the dataset, and SUVs are the most common body type.
# A few high-priced outliers were identified, but the overall data distribution provides meaningful insights into the used car market.
```
