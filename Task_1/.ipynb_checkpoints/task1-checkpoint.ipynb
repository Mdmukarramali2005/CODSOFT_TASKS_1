```python
import pandas as pd

df = pd.read_csv("data/automobile_dataset.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
numeric_cols = ["Engine_Size", "Horsepower", "Torque", "Accident_History", "Fuel_Efficiency"]

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
object_cols = ["Transmission", "Service_History", "Color", "Location"]

for col in object_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.info()
```

```python
df.to_csv("cleaned_car_dataset.csv", index=False)
```
